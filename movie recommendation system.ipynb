```python
# ============================================
# Movie Recommendation System
# Content-Based Filtering using TF-IDF
# ============================================

import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# -----------------------------
# Create Movie Dataset
# -----------------------------
movies = {
    'Movie': [
        'Avatar', 'Titanic', 'The Avengers', 'Iron Man', 'Captain America',
        'Thor', 'Doctor Strange', 'Black Panther', 'Spider-Man', 'Batman',
        'Superman', 'Wonder Woman', 'Aquaman', 'The Dark Knight', 'Joker',
        'Inception', 'Interstellar', 'The Matrix', 'John Wick',
        'Mission Impossible', 'Fast & Furious', 'Toy Story',
        'Finding Nemo', 'Frozen', 'Moana', 'Coco',
        'The Lion King', 'Kung Fu Panda', 'Shrek', 'Cars'
    ],

    'Genre': [
        'Action Adventure Sci-Fi',
        'Romance Drama',
        'Action Adventure Superhero',
        'Action Sci-Fi Superhero',
        'Action Adventure Superhero',
        'Fantasy Adventure Superhero',
        'Fantasy Adventure Superhero',
        'Action Adventure Superhero',
        'Action Adventure Superhero',
        'Action Crime Superhero',
        'Action Adventure Superhero',
        'Fantasy Action Adventure',
        'Adventure Fantasy Action',
        'Crime Action Thriller',
        'Crime Drama Thriller',
        'Sci-Fi Thriller',
        'Sci-Fi Adventure Drama',
        'Sci-Fi Action',
        'Action Crime Thriller',
        'Action Spy Adventure',
        'Action Racing Adventure',
        'Animation Family Comedy',
        'Animation Family Adventure',
        'Animation Musical Family',
        'Animation Adventure Family',
        'Animation Adventure Family',
        'Animation Adventure Drama',
        'Animation Comedy Adventure',
        'Animation Comedy Fantasy',
        'Animation Family Comedy'
    ]
}

df = pd.DataFrame(movies)

print(df.head())

# -----------------------------
# TF-IDF Vectorization
# -----------------------------
tfidf = TfidfVectorizer(stop_words='english')

tfidf_matrix = tfidf.fit_transform(df['Genre'])

# -----------------------------
# Cosine Similarity
# -----------------------------
cosine_sim = cosine_similarity(tfidf_matrix)

# -----------------------------
# Recommendation Function
# -----------------------------
def recommend(movie_name, top_n=10):

    movie_name = movie_name.lower()

    movie_list = df['Movie'].str.lower()

    if movie_name not in movie_list.values:
        print("Movie not found!")
        return

    index = movie_list[movie_list == movie_name].index[0]

    similarity_scores = list(enumerate(cosine_sim[index]))

    similarity_scores = sorted(similarity_scores,
                               key=lambda x: x[1],
                               reverse=True)

    similarity_scores = similarity_scores[1:top_n+1]

    print("\nRecommended Movies\n")

    for i, score in similarity_scores:
        print(f"{df.iloc[i]['Movie']}   ({round(score,2)})")

# -----------------------------
# User Input
# -----------------------------
movie = input("Enter Movie Name: ")

recommend(movie)
```
